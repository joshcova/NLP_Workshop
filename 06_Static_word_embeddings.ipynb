## Static word embedding

In this Jupyter notebook, we will examine how word embeddings work. We will be using some familiar and some new Python libraries.

By the end of this notebook, you will have developed a feeling of how word embeddings can be leveraged to answer social science research questions.

```python
# import or install the relevant libraries

import spacy
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Download the medium-sized English language model for spaCy (We have already used it for named entity
# recognition and part-of-speech tagging!)
# But today we will focus on word embeddings
!python -m spacy download en_core_web_md
```

```python
# Load the medium-sized English spaCy model
nlp = spacy.load("en_core_web_md")
```

```python
word = nlp.vocab["democracy"]
```

```python
# What does the shape correspond to?
word.vector.shape
```

```python
# Some words are out-of-vocabulary and will have zero vectors
# Iterate through a list of terms
for term in ["democracy", "authoritarianism", "populism", "xyzabc123", "Bundestag"]:
    tok = nlp.vocab[term]
    dimensions = tok.vector[:10].round(3)
    print(term, dimensions)
```

```python
vector_1 = nlp.vocab['banana'].vector
vector_2 = nlp.vocab['laptop'].vector

# Reshape the vectors to be 2D arrays as required by cosine_similarity
vector_1_reshaped = vector_1.reshape(1, -1);
vector_2_reshaped = vector_2.reshape(1, -1);

similarity = cosine_similarity(vector_1_reshaped, vector_2_reshaped)

# What is the substantive meaning of this?
```

```python
# Install the gensim library
# This is a GloVe model trained on a large corpus (Wikipedia and Gigaword) and
# provides 100-dimensional vectors. It's specifically designed for word embeddings.
!pip install gensim
```

```python
# Import the downloader utility from gensim.api
import gensim.downloader as api

# Load the 'glove-wiki-gigaword-100' pre-trained word embedding model (100 dimensions)
model = api.load("glove-wiki-gigaword-100")
```

```python
model["apple"].shape
```

```python
model.most_similar(positive="woman")
```

```python
# Perform a word analogy: find words most similar to (woman + king - man)
result = model.most_similar(
    positive=["woman", "king"],
    negative=["man"],
    topn=5
)


for word, score in result:
    print(word,score)
```

```python
# Perform a word analogy: find the top 1 word most similar to (woman + doctor - man)
model.most_similar(positive=["woman", "doctor"], negative=["man"], topn=1)
```

```python
# Use the 'doesnt_match' function to identify the word that is least similar to the others in the list
model.doesnt_match(["breakfast", "lunch", "dinner", "cereal"])   # Expected output: 'cereal'
```

An important problem is that of the entrenchment of algorithmic biases. Why does this happen?

```python
# Define a list of professions to analyze
professions = [
    "doctor", "programmer", "engineer", "scientist", "king",
    "ceo", "lawyer", "pilot", "nurse", "teacher"
]
```

```python
for job in professions:
    # Find the word most similar to (man + job - woman) - attempting to find the male-associated word
    male   = model.most_similar(positive=["man",   job], negative=["woman"], topn=1)[0][0]
    # Find the word most similar to (woman + job - man) - attempting to find the female-associated word
    female = model.most_similar(positive=["woman", job], negative=["man"],   topn=1)[0][0]
    # Print the profession and its male and female associated words
    print(job, male, female)
```
