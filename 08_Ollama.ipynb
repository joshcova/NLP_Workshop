````python
# once you have created a folder, created a csv file and have located all of this on VS Code, you can adapt this
# code

pip install ollama

import pandas as pd
import requests
import ollama

df = pd.read_csv(".csv")

def analyze_sentiment(text):
    prompt = f"""
Classify the following text as X, Y, or Z.

Text: "{text}"

Classification:
"""
    response = ollama.generate(
        model=" ", # model that you have downloaded
        prompt=prompt
    )

    return response["response"].strip()
````

````python
df["classificaation"] = df["text"].apply(analyze_sentiment) # apply the function
````

````python
# Last step!  Save the output:

df.to_csv(".csv", index=False)
````
